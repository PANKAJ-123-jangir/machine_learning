```python
#task-4
#tip_calculater
def tip_cal(bill,tip_per):         
    tip=(tip_per/100)*bill
    total=bill+tip
    return{
        "BILL":bill,"Tip":tip,"Toatal_Amount":total
    }
bill=float(input("enter the bill amount"))

tip_per=float(input("enter the tip percent"))

result=tip_cal(bill,tip_per)
print(result)
```

```python
students=[{"name": "oppo","score":59},
          {"name": "vivo","score":69},
          {"name": "samsung","score":79},
          {"name": "oneplus","score":89},
          {"name": "iphone","score":93}]

def classify(score):
    if score>=90:
        return "A"
    elif score>=75:
        return "B"
    elif score>=60:
        return "C"
    elif score>=40:
        return "D"
    else:
        return "F"

sorted_stu = sorted(students,key=lambda student: student["score"],reverse=True)

for student in sorted_stu:
    print(student["name"], "->", classify(student["score"]))
```
