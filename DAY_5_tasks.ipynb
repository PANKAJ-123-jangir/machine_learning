```python
#type hinted calculator
#task-4
def add(a: float, b: float) -> float:
   
    return a + b


def subtract(a: float, b: float) -> float:
   
    return a - b


def multiply(a: float, b: float) -> float:
  
    return a * b


def divide(a: float, b: float) -> float:
    
    if b == 0:
        print("Error: Cannot divide by zero")
        return 0

    return a / b


def power(base: float, exp: float) -> float:
   
    return base ** exp


def modulus(a: int, b: int) -> int:
    
    if b == 0:
        print("Error: Cannot perform modulus by zero")
        return 0

    return a % b


num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

op = input("Enter operation (+,-,*,/,**,%) : ")

if op == "+":
    print(add(num1, num2))

elif op == "-":
    print(subtract(num1, num2))

elif op == "*":
    print(multiply(num1, num2))

elif op == "/":
    print(divide(num1, num2))

elif op == "**":
    print(power(num1, num2))

elif op == "%":
    print(modulus(int(num1), int(num2)))

else:
    print("Invalid operator")
```

```python
#task 1
#class based student report

class Student:

    def __init__(self, name: str, roll_no: int, marks: list):

        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def average(self) -> float:
    
        return sum(self.marks) / len(self.marks)

    def grade(self) -> str:

        avg = self.average()

        if avg >= 90:
            return "A"
        elif avg >= 80:
            return "B"
        elif avg >= 70:
            return "C"
        elif avg >= 60:
            return "D"
        elif avg >= 50:
            return "E"
        else:
            return "F"

    def __str__(self) -> str:

        return (f"Name: {self.name}, "
                f"Roll No: {self.roll_no}, "
                f"Average: {self.average():.2f}, "
                f"Grade: {self.grade()}")


# Create 3 student objects

s1 = Student("Pankaj", 101, [90, 85, 88])

s2 = Student("Rahul", 102, [75, 80, 70])

s3 = Student("Aman", 103, [95, 92, 98])


# Print report cards

print(s1)
print(s2)
print(s3)
```

```python
#task-5
#inheritance library system
# means:
#A child class can reuse the properties and methods of a parent class.

class LibraryItem:
    def __init__(self,title,author,year):
        self.title=title
        self.author=author
        self.year=year
    def describe(self):
         print(f"Title: {self.title}, Author: {self.author}, Year: {self.year}")
         
class Book(LibraryItem):
    def __init__(self, title, author, year, pages):
        super().__init__(title, author, year)
        self.pages = pages
    def describe(self):
        print(f"Title: {self.title}, Author: {self.author}, Year: {self.year}, Pages: {self.pages}")
        
        
class Ebook(LibraryItem):
    def __init__(self, title, author, year, file_size_mb):
        super().__init__(title, author, year)
        self.file_size_mb = file_size_mb
    def describe(self):
        print(f"Title: {self.title}, Author: {self.author}, Year: {self.year}, File Size: {self.file_size_mb} MB")        

# Create objects

b1 = Book("Python Basics", "John Smith", 2020, 350)

b2 = Book("Machine Learning", "Andrew", 2023, 500)

e1 = Ebook("Data Science Guide", "Alice", 2022, 15)

e2 = Ebook("AI Handbook", "Bob", 2024, 25)


# Store in one list

library = [b1, b2, e1, e2]


# Loop through and call describe()

for item in library:

    item.describe()


#print(isinstance(my_book, LibraryItem
#isinstance() is used to check:  "Is this object an instance of a particular class?"    
#it prints the True
#Book inherits from LibraryItem.
#So every Book object is also considered a LibraryItem.
```

```python
#dunder methods 
#task-8

from math import gcd

class Fraction:

    def __init__(self, numerator: int, denominator: int):

        self.numerator = numerator
        self.denominator = denominator

    def __str__(self):

        return f"{self.numerator}/{self.denominator}"

    def __add__(self, other):

        num = (self.numerator * other.denominator) + \
              (other.numerator * self.denominator)

        den = self.denominator * other.denominator

        common = gcd(num, den)

        return Fraction(num // common, den // common)

    def __eq__(self, other):

        g1 = gcd(self.numerator, self.denominator)
        g2 = gcd(other.numerator, other.denominator)

        n1 = self.numerator // g1
        d1 = self.denominator // g1

        n2 = other.numerator // g2
        d2 = other.denominator // g2

        return n1 == n2 and d1 == d2

    def __lt__(self, other):

        return self.numerator * other.denominator < \
               other.numerator * self.denominator


# Test Pair 1

f1 = Fraction(1, 2)
f2 = Fraction(1, 3)

print("Addition:", f1 + f2)
print("Equal:", f1 == f2)
print("Less Than:", f1 < f2)

print()

# Test Pair 2

f3 = Fraction(2, 4)
f4 = Fraction(1, 2)

print("Addition:", f3 + f4)
print("Equal:", f3 == f4)
print("Less Than:", f3 < f4)

print()

# Test Pair 3

f5 = Fraction(3, 4)
f6 = Fraction(5, 6)

print("Addition:", f5 + f6)
print("Equal:", f5 == f6)
print("Less Than:", f5 < f6)
```

```python
#task-9
#mini inventory system

import csv

class Product:

    def __init__(self, name: str, price: float, quantity: int):

        self.name = name
        self.price = price
        self.quantity = quantity

    def __str__(self):

        return f"{self.name} | ₹{self.price} | Qty: {self.quantity}"


class Inventory:

    def __init__(self):

        self.products = []

    def add_product(self, product):

        self.products.append(product)

    def total_value(self) -> float:

        return sum(
            p.price * p.quantity
            for p in self.products
        )

    def find_product(self, name: str):

        for product in self.products:

            if product.name.lower() == name.lower():

                return product

        return None

    def save_to_csv(self, filename):

        with open(filename, "w", newline="") as file:

            writer = csv.writer(file)

            writer.writerow(
                ["name", "price", "quantity"]
            )

            for product in self.products:

                writer.writerow([
                    product.name,
                    product.price,
                    product.quantity
                ])

    def load_from_csv(self, filename):

        self.products.clear()

        with open(filename, "r") as file:

            reader = csv.DictReader(file)

            for row in reader:

                product = Product(
                    row["name"],
                    float(row["price"]),
                    int(row["quantity"])
                )

                self.products.append(product)


# Testing

inventory = Inventory()

inventory.add_product(
    Product("Laptop", 60000, 2)
)

inventory.add_product(
    Product("Mouse", 500, 10)
)

inventory.add_product(
    Product("Keyboard", 1500, 5)
)

print("Total Inventory Value:")
print(inventory.total_value())

print("\nSearch Product:")
print(inventory.find_product("mouse"))

inventory.save_to_csv("products.csv")

print("\nSaved to CSV")

new_inventory = Inventory()

new_inventory.load_from_csv("products.csv")

print("\nLoaded Products:")

for product in new_inventory.products:

    print(product)
```

```python
#explore
# What does @staticmethod do?
# A static method belongs to the class but does not use self or cls.
# It behaves like a normal function placed inside a class for organization.

# Could load_from_csv() be a static method?
# Yes. A static method can read the CSV file and create an Inventory object manually.
# Example: inventory = Inventory.load_from_csv("products.csv")

# Could load_from_csv() be a class method?
# Yes. A class method receives cls automatically and can create objects using cls().
# This is useful if Inventory is later inherited by another class.

# Difference:
# @staticmethod -> no self, no cls, cannot access class information automatically.
# @classmethod -> receives cls, can create and return objects of the current class.

# Conclusion:
# load_from_csv() is better as a class method because it creates and returns
# an Inventory object, making the code more flexible and OOP-friendly.
```
