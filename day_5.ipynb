```python
# ==========================================
# STACK IN PYTHON
# ==========================================

# Stack follows:
# LIFO -> Last In First Out

# Creating an empty stack using list
stack=[]

# PUSH operation using append()
stack.append(10)
stack.append(20)
stack.append(30)

# Printing stack
print("Stack after push operations:")
print(stack)

# POP operation
# Removes last inserted element
stack.pop()

print("Stack after pop operation:")
print(stack)

# Accessing top element
print("Top element of stack:")
print(stack[-1])



# ==========================================
# REVERSE STRING USING STACK
# ==========================================

# Original string
s="hello"

# Empty stack
stack=[]

# Push each character into stack
for ch in s:
    stack.append(ch)

# Empty string for reversed result
rev=""

# Pop characters one by one
while stack:
    rev+=stack.pop()

print("Reversed string:")
print(rev)



# ==========================================
# QUEUE IN PYTHON
# ==========================================

# Queue follows:
# FIFO -> First In First Out

# Import deque
from collections import deque

# Creating empty queue
q=deque()

# ENQUEUE operation using append()
q.append(10)
q.append(20)
q.append(30)

print("Queue after enqueue operations:")
print(q)

# DEQUEUE operation
# Removes first inserted element
q.popleft()

print("Queue after dequeue operation:")
print(q)

# Accessing front element
print("Front element of queue:")
print(q[0])



# ==========================================
# NESTED LIST
# ==========================================

# List inside another list
matrix=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

print("Nested List:")
print(matrix)

# Accessing element
# matrix[row][column]
print("Element at row 2 column 3:")
print(matrix[1][2])



# ==========================================
# LOOP THROUGH NESTED LIST
# ==========================================

print("Printing all elements of nested list:")

for row in matrix:
    for val in row:
        print(val,end=" ")

print()



# ==========================================
# NESTED DICTIONARY
# ==========================================

# Dictionary inside dictionary
students={
    "pankaj":{"age":20,"marks":90},
    "rahul":{"age":21,"marks":85}
}

print("Nested Dictionary:")
print(students)

# Accessing nested value
print("Marks of Pankaj:")
print(students["pankaj"]["marks"])



# ==========================================
# LIST OF DICTIONARIES
# ==========================================

# Very important in ML and JSON data
students_list=[
    {"name":"Pankaj","marks":90},
    {"name":"Rahul","marks":85}
]

print("List of Dictionaries:")
print(students_list)

# Accessing values
print("First student name:")
print(students_list[0]["name"])

```

```python
# ==========================================
# CSV FILE HANDLING IN PYTHON
# ==========================================

# CSV -> Comma Separated Values

# Import csv module
import csv

# ------------------------------------------
# WRITING INTO CSV FILE
# ------------------------------------------

# Data to write
data=[
    ["Name","Marks","City"],
    ["Pankaj",90,"Jaipur"],
    ["Rahul",85,"Delhi"],
    ["Aman",88,"Mumbai"]
]

# Creating and writing csv file
with open("students.csv","w",newline="") as file:

    writer=csv.writer(file)

    # Writing multiple rows
    writer.writerows(data)

print("CSV file created successfully")



# ------------------------------------------
# READING CSV FILE
# ------------------------------------------

with open("students.csv","r") as file:

    reader=csv.reader(file)

    # Reading row by row
    for row in reader:
        print(row)



# ==========================================
# JSON FILE HANDLING IN PYTHON
# ==========================================

# JSON -> JavaScript Object Notation

# Import json module
import json

# ------------------------------------------
# WRITING JSON FILE
# ------------------------------------------

student={
    "name":"Pankaj",
    "age":20,
    "marks":90,
    "skills":["Python","C++","ML"]
}

# Writing json data
with open("student.json","w") as file:

    json.dump(student,file,indent=4)

print("JSON file created successfully")



# ------------------------------------------
# READING JSON FILE
# ------------------------------------------

with open("student.json","r") as file:

    data=json.load(file)

    print(data)

    print("Student Name:")
    print(data["name"])



# ==========================================
# EXCEL FILE HANDLING IN PYTHON
# ==========================================

# Need pandas library
# Install using:
# pip install pandas openpyxl

import pandas as pd

# ------------------------------------------
# WRITING EXCEL FILE
# ------------------------------------------

data={
    "Name":["Pankaj","Rahul","Aman"],
    "Marks":[90,85,88],
    "City":["Jaipur","Delhi","Mumbai"]
}

# Creating DataFrame
df=pd.DataFrame(data)

# Writing excel file
df.to_excel("students.xlsx",index=False)

print("Excel file created successfully")



# ------------------------------------------
# READING EXCEL FILE
# ------------------------------------------

# Reading excel file
df=pd.read_excel("students.xlsx")

print(df)



# ==========================================
# IMPORTANT POINTS
# ==========================================

# CSV:
# - Stores tabular data
# - Values separated by commas
# - Uses csv module

# JSON:
# - Stores structured data
# - Used in APIs and web development
# - Uses json module

# Excel:
# - Spreadsheet file
# - Uses pandas library
# - Very important in Data Science and ML



# ==========================================
# REAL WORLD USES
# ==========================================

# CSV:
# Dataset storage

# JSON:
# APIs
# Web applications

# Excel:
# Data analysis
# Machine Learning datasets
# Reports
```
