```python
import numpy as np
#creating a numpy array
arr=np.array([1,2,3,4,5])
print(arr)
#checking type
print(type(arr))
```

```python
#array addtion
print(2*arr)
print(arr*2)
# for list
li=[1,2,3,4]
print(arr*2)
print(2*arr)
#in allabove cases outputs are same
```

```python
# some mathematical prebuild functions in numpy
# for arr=np.array([1,2,3,4,5])
print(np.sum(arr))
print(np.max(arr))
print(np.min(arr))
print(np.mean(arr))
print("these are soome direct numpy opeartions on arrays")
```

```python
arr2=np.array([[1,2,3],
               [1,2,3],
               [1,2,3]])
print(arr2)
print(arr2.shape)  #shape tell us about matrix order
```

```python
# zeroes and ones array
print(np.zeros((2,4)))

print(np.ones((2,3)))
```

```python
#arrange array
arr3=np.arange(1,10)
print(arr3)
#reshape in any 2D array we want
arr4=arr3.reshape(3,3)
print(arr4)
```

slicing and indexing same as normal array

```python
#creating empty 2D array
#better to use zero array
emp_arr=np.empty((3,3))
print(emp_arr)
emp_arr[0][0]=1
print(emp_arr)
emp_arr=np.zeros((3,3))
print(emp_arr)
```

```python
#printing random 3X3 matrix from range 1---100
import numpy as np
ar=np.random.randint(1,100,(3,3))
print(ar)
```

```python
#flatten array ---2d --1d
print(arr4.flatten())
print(arr4)
```

```python
#broadcasting
#doing opeartions on each elemenst on array
```

```python
#boolean indexing
print(arr>3)
#print true on value greater than 3 anf false on left indexes
```

```python
#filter measn if we want to print elements greagter than a value
print(arr[arr>3])
```

```python
#fancy indexing
#we can print any element we want without loop
print(arr[[1,3,4]])
```

```python
#axis concept
#print sum row wise and column wise
#axis=1  for row
#axis=0 for column
print(arr4)
print(np.sum(arr4,axis=1))
```

```python
#print col wise
print(np.sum(arr4,axis=0))
```

```python
#print all element sum
print(np.sum(arr4,axis=None))
```

```python
#specific row sum
#2nd r0w sum
print(np.sum(arr4[1,:]))
```

```python
#first col sum
print(np.sum(arr4[:,0]))
```

```python
#stacking of arrays
#combining varoius 1D array into 2D array row wise stack
a=np.array([1,2,3])
b=np.array([4,5,6])

print(np.vstack((a,b)))
#in 1D array stacking
print(np.hstack((a,b)))
```
